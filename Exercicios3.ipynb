```python
def buscar_cliente(nome, lista_clientes):
    for i in range(len(lista_clientes)):
        if lista_clientes[i] == nome:
            return i  # encontrou, retorna a posição
    return -1  # não encontrou, retorna -1


# Lista de exemplo e testes da função
clientes = ["Ana", "Bruno", "Carlos", "Daniel", "Elisa"]
busca1 = "Carlos"
busca2 = "Fernanda"

# Teste com busca1 (Carlos)
pos1 = buscar_cliente(busca1, clientes)
if pos1 != -1:
    print(f"Cliente '{busca1}' encontrado na posição {pos1}.")
else:
    print(f"Cliente '{busca1}' não encontrado na lista.")

# Teste com busca2 (Fernanda)
pos2 = buscar_cliente(busca2, clientes)
if pos2 != -1:
    print(f"Cliente '{busca2}' encontrado na posição {pos2}.")
else:
    print(f"Cliente '{busca2}' não encontrado na lista.")
    
```

```python
def conta_ocorrencias(valor, lista):
    contador = 0
    for elemento in lista:
        if elemento == valor:
            contador += 1
    return contador


# Testes da função
print(conta_ocorrencias(3, [1, 3, 5, 3, 7]))  # Saída esperada: 2
print(conta_ocorrencias("a", ["a", "b", "c", "a"]))  # Saída esperada: 2
print(
    conta_ocorrencias(10, [11, 12, 13])
)  # Saída esperada: 0 (10 não está na lista)
```

```python
def remove_elemento(lista, valor):
    if len(lista) == 0:
        return None  # Lista vazia, nada a remover

    # Busca sequencial pelo valor
    indice = -1
    for i in range(len(lista)):
        if lista[i] == valor:
            indice = i
            break

    if indice == -1:
        return None  # Valor não encontrado na lista

    # Se encontrou, vamos remover deslocando os elementos
    elemento_removido = lista[indice]

    # Desloca os elementos à direita do indice para a esquerda
    for j in range(indice, len(lista) - 1):
        lista[j] = lista[j + 1]

    lista.pop()  # remove o último elemento duplicado após o deslocamento
    return elemento_removido


# Exemplo de uso
dados = [10, 3, 7, 34, 23, 2, 21]
print("Lista original:", dados)

rem = remove_elemento(dados, 23)

if rem is not None:
    print(f"Valor {rem} removido.")
else:
    print("Valor não encontrado ou lista vazia.")

print("Lista após remoção:", dados)
```

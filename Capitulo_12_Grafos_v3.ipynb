```python
G = {
    'A': ['B', 'C'],
    'B': ['A', 'D', 'E'],
    'C': ['A', 'F'],
    'D': ['B'],
    'E': ['B', 'F'],
    'F': ['C', 'E']
}

def dfs(grafo, inicio, visitados=None):
    if visitados is None:
        visitados = []
    visitados.append(inicio)
    for vizinho in grafo[inicio]:
        if vizinho not in visitados:
            dfs(grafo, vizinho, visitados)
    return visitados

ordem = dfs(G, 'A')
print(ordem)
```

```python
from collections import deque

def bfs(grafo, inicio):
    visitados = [inicio]
    fila = deque([inicio])
    ordem = []
    while fila:
        vertice = fila.popleft()
        ordem.append(vertice)
        for vizinho in grafo[vertice]:
            if vizinho not in visitados:
                visitados.append(vizinho)
                fila.append(vizinho)
    return ordem

ordem_bfs = bfs(G, 'A')
print(ordem_bfs)
```

```python
import networkx as nx

G_cidades = nx.Graph()
G_cidades.add_weighted_edges_from([
    ('São Paulo', 'Campinas', 95),
    ('São Paulo', 'Sorocaba', 87),
    ('Sorocaba', 'Campinas', 85),
    ('Sorocaba', 'Botucatu', 152),
    ('Botucatu', 'Jaú', 76),
    ('Jaú', 'São Carlos', 99),
    ('Campinas', 'Limeira', 54),
    ('Limeira', 'São Carlos', 98),
    ('Campinas', 'Ribeirão Preto', 238),
    ('Ribeirão Preto', 'Barretos', 127),
    ('São Carlos', 'S. J. Rio Preto', 216),
    ('São Carlos', 'Barretos', 226),
    ('S. J. Rio Preto', 'Barretos', 95)
])

origem = 'São Paulo'
destino = 'Barretos'

menor_caminho = nx.shortest_path(G_cidades, origem, destino, weight='weight')
distancia = nx.shortest_path_length(G_cidades, origem, destino, weight='weight')

print('Menor caminho:', menor_caminho)
print('Distância total:', distancia, 'km')
```

```python
from collections import deque

rede = {
    'A': ['B', 'D'],
    'B': ['A', 'C'],
    'C': ['B', 'D', 'E'],
    'D': ['A', 'C', 'F'],
    'E': ['C'],
    'F': ['D']
}

def bfs_caminho(grafo, inicio, destino):
    visitados = {inicio}
    fila = deque([(inicio, [inicio])])
    while fila:
        vertice, caminho = fila.popleft()
        if vertice == destino:
            return caminho
        for vizinho in grafo[vertice]:
            if vizinho not in visitados:
                visitados.add(vizinho)
                fila.append((vizinho, caminho + [vizinho]))
    return None

origem = 'A'
destino = 'E'
caminho = bfs_caminho(rede, origem, destino)

print(f"Caminho mínimo de {origem} até {destino}: {caminho}")
print('Graus de separação:', len(caminho) - 1)
```

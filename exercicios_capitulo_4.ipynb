### Exercício 3 (PRÁTICA)
**Enunciado:** Implemente a função `busca_binaria` de forma recursiva em Python.

```python
def busca_binaria_recursiva(vetor, alvo, esq=0, dir=None):
    if dir is None:
        dir = len(vetor) - 1
        
    if esq > dir:
        return -1 # Caso base: elemento não encontrado
        
    meio = (esq + dir) // 2
    
    if vetor[meio] == alvo:
        return meio # Caso base: elemento encontrado
    elif vetor[meio] < alvo:
        return busca_binaria_recursiva(vetor, alvo, meio + 1, dir)
    else:
        return busca_binaria_recursiva(vetor, alvo, esq, meio - 1)

# Teste
vetor_teste = [10, 20, 30, 40, 50, 60, 70]
print("Índice do valor 50:", busca_binaria_recursiva(vetor_teste, 50))
print("Índice do valor 25:", busca_binaria_recursiva(vetor_teste, 25))
```

### Exercício 4 (PRÁTICA)
**Enunciado:** Considere uma lista de 1000 números inteiros ordenados. Compare a busca sequencial e a busca binária contando o número de comparações para o pior caso (elemento não presente).

```python
def busca_sequencial_contando(lista, alvo):
    comparacoes = 0
    for elemento in lista:
        comparacoes += 1
        if elemento == alvo:
            return -1, comparacoes
    return -1, comparacoes

def busca_binaria_contando(lista, alvo):
    comparacoes = 0
    esq, dir = 0, len(lista) - 1
    while esq <= dir:
        comparacoes += 1
        meio = (esq + dir) // 2
        if lista[meio] == alvo:
            return meio, comparacoes
        elif lista[meio] < alvo:
            esq = meio + 1
        else:
            dir = meio - 1
    return -1, comparacoes

# Teste comparativo com 1000 elementos
lista_1000 = list(range(0, 2000, 2)) # 1000 números pares de 0 a 1998
alvo_inexistente = 1999 # Número ímpar que obriga o pior caso

_, comp_seq = busca_sequencial_contando(lista_1000, alvo_inexistente)
_, comp_bin = busca_binaria_contando(lista_1000, alvo_inexistente)

print(f"Total de comparações - Busca Sequencial: {comp_seq}")
print(f"Total de comparações - Busca Binária: {comp_bin}")
```

### Projeto 1: Busca Eficiente em um Catálogo de Produtos Online
**Contexto:** Catálogo ordenado alfabeticamente pelo nome do produto. Implementar busca binária em lista de dicionários.

```python
# Projeto 1: Catálogo de Produtos Online

catalogo_produtos = [
    {"nome": "Camiseta Algodão", "preco": 49.90, "descricao": "Camiseta 100% algodão confortável"},
    {"nome": "Fone de Ouvido Bluetooth", "preco": 199.90, "descricao": "Fone sem fio com cancelamento de ruído"},
    {"nome": "Mochila Executiva", "preco": 150.00, "descricao": "Mochila resistente para notebook"},
    {"nome": "Notebook Pro 15", "preco": 4500.00, "descricao": "Intel i7, 16GB RAM, SSD 512GB"},
    {"nome": "Smartphone X", "preco": 2999.00, "descricao": "Tela OLED de 6.5 polegadas"},
    {"nome": "Teclado Mecânico", "preco": 250.00, "descricao": "Teclado RGB switch azul"}
] # Já ordenado pelo campo 'nome'

def buscar_produto(catalogo, nome_produto):
    esq, dir = 0, len(catalogo) - 1
    nome_alvo = nome_produto.lower()
    
    while esq <= dir:
        meio = (esq + dir) // 2
        nome_meio = catalogo[meio]["nome"].lower()
        
        if nome_meio == nome_alvo:
            return catalogo[meio]
        elif nome_meio < nome_alvo:
            esq = meio + 1
        else:
            dir = meio - 1
            
    return None

# Teste do Projeto 1
produto = buscar_produto(catalogo_produtos, "Notebook Pro 15")
if produto:
    print("--- Produto Encontrado ---")
    print(f"Nome: {produto['nome']} | Preço: R${produto['preco']:.2f} | Descrição: {produto['descricao']}")
else:
    print("Produto não encontrado no catálogo.")
```

### Projeto 2: Busca Eficiente em um Índice de Livros
**Contexto:** Índice de biblioteca mantido em ordem alfabética pelo título.

```python
# Projeto 2: Índice de Livros em Biblioteca

indice_livros = [
    {"titulo": "1984", "autor": "George Orwell", "ano": 1949},
    {"titulo": "Algoritmos: Teoria e Prática", "autor": "Thomas H. Cormen", "ano": 2012},
    {"titulo": "Dom Casmurro", "autor": "Machado de Assis", "ano": 1899},
    {"titulo": "O Hobit", "autor": "J.R.R. Tolkien", "ano": 1937},
    {"titulo": "O Pequeno Príncipe", "autor": "Antoine de Saint-Exupéry", "ano": 1943}
] # Ordenado por 'titulo'

def buscar_livro(indice, titulo_buscado):
    esq, dir = 0, len(indice) - 1
    alvo = titulo_buscado.lower()
    
    while esq <= dir:
        meio = (esq + dir) // 2
        titulo_meio = indice[meio]["titulo"].lower()
        
        if titulo_meio == alvo:
            return indice[meio]
        elif titulo_meio < alvo:
            esq = meio + 1
        else:
            dir = meio - 1
            
    return None

# Teste do Projeto 2
livro = buscar_livro(indice_livros, "Dom Casmurro")
if livro:
    print("--- Livro Encontrado ---")
    print(f"Título: {livro['titulo']} | Autor: {livro['autor']} | Ano: {livro['ano']}")
else:
    print("Livro não encontrado no acervo.")
```

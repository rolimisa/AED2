### Exercício 5 (PRÁTICA)
**Enunciado:** Classe `Pessoa` com `nome` e `idade`, método `faz_aniversario()` e `__str__()`.

```python
class Pessoa:
    def __init__(self, nome, idade):
        self.nome = nome
        self.idade = idade
        
    def faz_aniversario(self):
        self.idade += 1
        
    def __str__(self):
        return f"{self.nome}, {self.idade} anos"

# Demonstração
p1 = Pessoa("Ana", 20)
p2 = Pessoa("Lucas", 25)

print("Antes do aniversário:", p1, "|", p2)
p1.faz_aniversario()
p2.faz_aniversario()
print("Depois do aniversário:", p1, "|", p2)
```

### Exercício 6 (PRÁTICA)
**Enunciado:** Encapsulamento na classe `Lampada` (atributo privado `__estado`, métodos `ligar()`, `desligar()`, `is_acesa()`).

```python
class Lampada:
    def __init__(self):
        self.__estado = False # Atributo privado
        
    def ligar(self):
        self.__estado = True
        
    def desligar(self):
        self.__estado = False
        
    def is_acesa(self):
        return self.__estado

# Teste
l = Lampada()
print("A lâmpada está acesa?", l.is_acesa())
l.ligar()
print("Após ligar:", l.is_acesa())
l.desligar()
print("Após desligar:", l.is_acesa())
```

### Exercícios 7 e 8 (PRÁTICA)
**Enunciado:** Herança e Polimorfismo com `Animal`, `Cachorro`, `Gato` e função `animais_falam()`.

```python
class Animal:
    def fazer_som(self):
        print("Algum som")

class Cachorro(Animal):
    def fazer_som(self):
        print("Au Au")

class Gato(Animal):
    def fazer_som(self):
        print("Miau")

# Função polimórfica (Exercício 8)
def animais_falam(lista_animais):
    for animal in lista_animais:
        animal.fazer_som()

# Testes
meu_cao = Cachorro()
meu_gato = Gato()

animais_falam([meu_cao, meu_gato, Animal()])
```

### Projeto 1: Biblioteca de Livros
**Contexto:** Sistema para gerenciar empréstimo de livros e cadastro de usuários.

```python
class Livro:
    def __init__(self, titulo, autor, qtd_disponivel):
        self.titulo = titulo
        self.autor = autor
        self.__qtd_disponivel = qtd_disponivel
        
    def get_qtd_disponivel(self):
        return self.__qtd_disponivel
        
    def emprestar(self):
        if self.__qtd_disponivel > 0:
            self.__qtd_disponivel -= 1
            return True
        return False
        
    def devolver(self):
        self.__qtd_disponivel += 1

class Usuario:
    def __init__(self, nome, matricula):
        self.nome = nome
        self.matricula = matricula
        self.livros_emprestados = []
        
    def pegar_emprestado(self, livro):
        if livro.emprestar():
            self.livros_emprestados.append(livro)
            print(f"{self.nome} emprestou '{livro.titulo}'.")
        else:
            print(f"Empréstimo falhou: '{livro.titulo}' está indisponível.")
            
    def devolver_livro(self, livro):
        if livro in self.livros_emprestados:
            livro.devolver()
            self.livros_emprestados.remove(livro)
            print(f"{self.nome} devolveu '{livro.titulo}'.")

# Teste Projeto 1
livro1 = Livro("Estruturas de Dados", "Autor A", 1)
u1 = Usuario("Carlos", "1001")
u2 = Usuario("Mariana", "1002")

u1.pegar_emprestado(livro1)
u2.pegar_emprestado(livro1) # Deve falhar por indisponibilidade
u1.devolver_livro(livro1)
u2.pegar_emprestado(livro1) # Agora deve funcionar
```

### Projeto 2: Gerenciamento de Conta Bancária com Herança
**Contexto:** Conta base com `ContaCorrente` e `ContaPoupanca`.

```python
class Conta:
    def __init__(self, titular, saldo_inicial=0.0):
        self.titular = titular
        self._saldo = saldo_inicial
        
    def depositar(self, valor):
        if valor > 0:
            self._saldo += valor
            print(f"Depósito de R${valor:.2f} realizado. Saldo: R${self._saldo:.2f}")
            
    def sacar(self, valor):
        if 0 < valor <= self._saldo:
            self._saldo -= valor
            print(f"Saque de R${valor:.2f} realizado. Saldo: R${self._saldo:.2f}")
            return True
        print("Saldo insuficiente para saque.")
        return False

class ContaCorrente(Conta):
    taxa_saque = 2.50
    
    def sacar(self, valor):
        valor_total = valor + self.taxa_saque
        print(f"[Conta Corrente] Tentativa de saque com taxa de R${self.taxa_saque:.2f}:")
        return super().sacar(valor_total)

class ContaPoupanca(Conta):
    taxa_juros = 0.05 # 5% ao ano
    
    def render_juros(self):
        juros = self._saldo * self.taxa_juros
        self._saldo += juros
        print(f"[Conta Poupança] Juros de R${juros:.2f} aplicados. Saldo: R${self._saldo:.2f}")

# Teste Projeto 2
cc = ContaCorrente("João", 100.0)
cp = ContaPoupanca("Maria", 1000.0)

cc.sacar(20.0)
cp.render_juros()
```

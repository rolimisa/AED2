```python
class Pessoa:
    def __init__(self, nome, idade):
        self.nome = nome
        self.idade = idade

    def faz_aniversario(self):
        self.idade += 1

    def __str__(self):
        return f"{self.nome}, {self.idade} anos"

p1 = Pessoa("Ana", 20)
p2 = Pessoa("Carlos", 30)

print("Antes:")
print(p1)
print(p2)

p1.faz_aniversario()
p2.faz_aniversario()

print("Depois:")
print(p1)
print(p2)
```

```python
class Lampada:
    def __init__(self):
        self.__estado = False

    def ligar(self):
        self.__estado = True

    def desligar(self):
        self.__estado = False

    def is_acesa(self):
        return self.__estado

lampada = Lampada()
print(lampada.is_acesa())
lampada.ligar()
print(lampada.is_acesa())
lampada.desligar()
print(lampada.is_acesa())
```

```python
class Animal:
    def fazer_som(self):
        print("Algum som")

class Cachorro(Animal):
    def fazer_som(self):
        print("Au Au")

class Gato(Animal):
    def fazer_som(self):
        print("Miau")

a = Animal()
c = Cachorro()
g = Gato()

a.fazer_som()
c.fazer_som()
g.fazer_som()
```

```python
def animais_falam(lista_animais):
    for animal in lista_animais:
        animal.fazer_som()

lista_animais = [Cachorro(), Gato(), Cachorro()]
animais_falam(lista_animais)
```

```python
class Livro:
    def __init__(self, titulo, autor, quantidade):
        self.__titulo = titulo
        self.__autor = autor
        self.__quantidade = quantidade

    def get_titulo(self):
        return self.__titulo

    def __str__(self):
        return f"{self.__titulo} por {self.__autor} (Disponíveis: {self.__quantidade})"

    def emprestar(self):
        if self.__quantidade > 0:
            self.__quantidade -= 1
            return True
        return False

    def devolver(self):
        self.__quantidade += 1

class Usuario:
    def __init__(self, nome, matricula):
        self.__nome = nome
        self.__matricula = matricula
        self.__livros_emprestados = []

    def __str__(self):
        return f"{self.__nome} ({self.__matricula})"

    def emprestar_livro(self, livro):
        if livro.emprestar():
            self.__livros_emprestados.append(livro)
            print(f"{self.__nome} emprestou '{livro.get_titulo()}'.")
        else:
            print(f"Livro '{livro.get_titulo()}' indisponível.")

    def devolver_livro(self, livro):
        if livro in self.__livros_emprestados:
            livro.devolver()
            self.__livros_emprestados.remove(livro)
            print(f"{self.__nome} devolveu '{livro.get_titulo()}'.")
        else:
            print(f"{self.__nome} não tinha '{livro.get_titulo()}' emprestado.")

    def listar_emprestimos(self):
        titulos = [livro.get_titulo() for livro in self.__livros_emprestados]
        print(f"Livros de {self.__nome}: {titulos}")

livro1 = Livro("O Pequeno Príncipe", "Saint-Exupéry", 3)
livro2 = Livro("1984", "George Orwell", 2)
usuario1 = Usuario("Alice", "2023A")
usuario2 = Usuario("Bruno", "2023B")

print(livro1)
print(livro2)

usuario1.emprestar_livro(livro1)
usuario1.emprestar_livro(livro2)
usuario2.emprestar_livro(livro1)
usuario2.emprestar_livro(livro1)
usuario2.emprestar_livro(livro1)
usuario2.emprestar_livro(livro1)

usuario1.listar_emprestimos()
usuario2.listar_emprestimos()

usuario1.devolver_livro(livro1)
usuario1.listar_emprestimos()
print(livro1)
```

```python
class Conta:
    def __init__(self, titular, saldo_inicial=0):
        self.__titular = titular
        self.__saldo = saldo_inicial

    def depositar(self, valor):
        if valor > 0:
            self.__saldo += valor
            print(f"Depósito de {valor:.2f} realizado em {self.__titular}.")
        else:
            print("Valor de depósito inválido.")

    def sacar(self, valor):
        if 0 < valor <= self.__saldo:
            self.__saldo -= valor
            print(f"Saque de {valor:.2f} realizado em {self.__titular}.")
        else:
            print("Saque não autorizado.")

    def get_saldo(self):
        return self.__saldo

    def __str__(self):
        return f"{self.__titular}: Saldo R${self.__saldo:.2f}"

class ContaCorrente(Conta):
    def __init__(self, titular, saldo_inicial=0, limite=0):
        super().__init__(titular, saldo_inicial)
        self.__limite = limite

    def sacar(self, valor):
        if 0 < valor <= self.get_saldo() + self.__limite:
            novo_saldo = self.get_saldo() - valor
            super()._Conta__saldo = novo_saldo
            if novo_saldo < 0:
                print(f"Saque de {valor:.2f} realizado (incluindo uso do limite).")
            else:
                print(f"Saque de {valor:.2f} realizado.")
        else:
            print("Saque não autorizado: limite excedido.")

class ContaPoupanca(Conta):
    juros = 0.05

    def render_juros(self):
        rendimento = self.get_saldo() * ContaPoupanca.juros
        super()._Conta__saldo += rendimento
        print(f"Juros de R${rendimento:.2f} aplicados.")

cc = ContaCorrente("Cliente CC", 1000.0, limite=500.0)
cp = ContaPoupanca("Cliente CP", 2000.0)

print(cc)
print(cp)

cc.sacar(1200)
print(cc)

cc.depositar(300)
print(cc)

cp.sacar(500)
print(cp)

cp.render_juros()
print(cp)
```

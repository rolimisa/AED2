```python
class No:
    def __init__(self, valor):
        self.valor = valor
        self.esquerda = None
        self.direita = None

raiz = No('a')
raiz.esquerda = No('b')
raiz.direita = No('c')
raiz.esquerda.esquerda = No('d')
raiz.esquerda.direita = No('e')

def altura(no):
    if no is None:
        return 0
    alt_e = altura(no.esquerda)
    alt_d = altura(no.direita)
    return 1 + max(alt_e, alt_d)

print(altura(raiz))
```

```python
def conta_folhas(no):
    if no is None:
        return 0
    if no.esquerda is None and no.direita is None:
        return 1
    return conta_folhas(no.esquerda) + conta_folhas(no.direita)

print(conta_folhas(raiz))
```

```python
from sklearn import tree

X = [
    [25, 5000],
    [40, 12000],
    [22, 3000],
    [35, 8000],
    [50, 10000]
]
y = [1, 1, 0, 1, 0]

clf = tree.DecisionTreeClassifier()
clf = clf.fit(X, y)

previsao = clf.predict([[30, 7000]])
print(previsao)
```

```python
familia = {
    'João': ['Pedro', 'Ana'],
    'Pedro': ['Carlos', 'Marina'],
    'Ana': ['Julia'],
    'Carlos': [],
    'Marina': [],
    'Julia': []
}

def exibir_arvore(nome, nivel=0):
    print('  ' * nivel + nome)
    for filho in familia.get(nome, []):
        exibir_arvore(filho, nivel + 1)

exibir_arvore('João')
```

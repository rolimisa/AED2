```python
def busca_binaria_recursiva(vetor, alvo, esq=0, dir=None):
    if dir is None:
        dir = len(vetor) - 1
    if esq > dir:
        return -1
    meio = (esq + dir) // 2
    if vetor[meio] == alvo:
        return meio
    elif vetor[meio] < alvo:
        return busca_binaria_recursiva(vetor, alvo, meio + 1, dir)
    else:
        return busca_binaria_recursiva(vetor, alvo, esq, meio - 1)

numeros = [2, 5, 8, 12, 16, 23, 38, 56, 72, 91]

print(busca_binaria_recursiva(numeros, 23))
print(busca_binaria_recursiva(numeros, 50))
```

```python
lista = list(range(1000))
alvo = 1001

comps_sequencial = 0
comps_binaria = 0

def busca_sequencial_contando(vetor, valor):
    global comps_sequencial
    for indice in range(len(vetor)):
        comps_sequencial += 1
        if vetor[indice] == valor:
            return indice
    return -1

def busca_binaria_contando(vetor, valor):
    global comps_binaria
    esq = 0
    dir = len(vetor) - 1
    while esq <= dir:
        meio = (esq + dir) // 2
        comps_binaria += 1
        if vetor[meio] == valor:
            return meio
        elif vetor[meio] < valor:
            esq = meio + 1
        else:
            dir = meio - 1
    return -1

busca_sequencial_contando(lista, alvo)
busca_binaria_contando(lista, alvo)

print(f"Comparações - Busca Sequencial: {comps_sequencial}")
print(f"Comparações - Busca Binária: {comps_binaria}")
```

```python
def buscar_produto_catalogo(catalogo, nome_produto):
    baixo = 0
    alto = len(catalogo) - 1
    alvo = nome_produto.lower()
    while baixo <= alto:
        meio = (baixo + alto) // 2
        nome_meio = catalogo[meio]["nome"].lower()
        if nome_meio == alvo:
            return catalogo[meio]
        elif nome_meio < alvo:
            baixo = meio + 1
        else:
            alto = meio - 1
    return None

catalogo_online = [
    {"nome": "Câmera", "preco": 1899.90, "descricao": "Câmera digital"},
    {"nome": "Fone", "preco": 249.90, "descricao": "Fone sem fio"},
    {"nome": "Notebook", "preco": 4599.00, "descricao": "Notebook 15 polegadas"},
    {"nome": "Smartphone", "preco": 2299.00, "descricao": "Smartphone 5G"},
    {"nome": "Tablet", "preco": 1599.00, "descricao": "Tablet 10 polegadas"}
]

nome_busca = "Notebook"
produto = buscar_produto_catalogo(catalogo_online, nome_busca)

if produto:
    print(f"Produto encontrado: {produto['nome']}")
    print(f"Preço: R${produto['preco']:.2f}")
    print(f"Descrição: {produto['descricao']}")
else:
    print(f"O produto '{nome_busca}' não foi encontrado no catálogo.")
```

```python
def buscar_livro_indice(indice_livros, titulo_livro):
    baixo = 0
    alto = len(indice_livros) - 1
    alvo = titulo_livro.lower()
    while baixo <= alto:
        meio = (baixo + alto) // 2
        titulo_meio = indice_livros[meio]["titulo"].lower()
        if titulo_meio == alvo:
            return indice_livros[meio]
        elif titulo_meio < alvo:
            baixo = meio + 1
        else:
            alto = meio - 1
    return None

indice_biblioteca = [
    {"titulo": "1984", "autor": "George Orwell", "ano": 1949},
    {"titulo": "Dom Casmurro", "autor": "Machado de Assis", "ano": 1899},
    {"titulo": "O Pequeno Príncipe", "autor": "Antoine de Saint-Exupéry", "ano": 1943},
    {"titulo": "Orgulho e Preconceito", "autor": "Jane Austen", "ano": 1813},
    {"titulo": "Vidas Secas", "autor": "Graciliano Ramos", "ano": 1938}
]

titulo_busca = "O Pequeno Príncipe"
livro = buscar_livro_indice(indice_biblioteca, titulo_busca)

if livro:
    print(f"Livro encontrado: {livro['titulo']}")
    print(f"Autor: {livro['autor']}")
    print(f"Ano de publicação: {livro['ano']}")
else:
    print(f"O livro com o título '{titulo_busca}' não foi encontrado na biblioteca.")
```

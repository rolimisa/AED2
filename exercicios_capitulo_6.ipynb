### Exercício 5 (PRÁTICA)
**Enunciado:** Função `decimal_para_hexa(n)` usando pilha.

```python
def decimal_para_hexa(n):
    if n == 0:
        return "0"
        
    digitos_hex = "0123456789ABCDEF"
    pilha = []
    
    while n > 0:
        resto = n % 16
        pilha.append(digitos_hex[resto])
        n = n // 16
        
    resultado = ""
    while pilha:
        resultado += pilha.pop()
        
    return resultado

# Teste
print("26 em hexadecimal:", decimal_para_hexa(26)) # Esperado: '1A'
print("255 em hexadecimal:", decimal_para_hexa(255)) # Esperado: 'FF'
```

### Exercício 6 (PRÁTICA)
**Enunciado:** Função `balanceia_colchetes(expressao)` usando pilha.

```python
def balanceia_colchetes(expressao):
    pilha = []
    mapeamento = {')': '(', '}': '{', ']': '['}
    
    for caractere in expressao:
        if caractere in "({[":
            pilha.append(caractere)
        elif caractere in ")}]":
            if not pilha:
                return False
            topo = pilha.pop()
            if topo != mapeamento[caractere]:
                return False
                
    return len(pilha) == 0

# Testes
print("({[]}) está balanceado?", balanceia_colchetes("({[]})"))
print("({[}]) está balanceado?", balanceia_colchetes("({[}])"))
```

### Projeto 1: Histórico de Filtros em Dashboard de Dados
**Contexto:** Mecanismo de desfazer (Undo) filtros aplicados em um dashboard usando pilha de estados.

```python
class DashboardFiltros:
    def __init__(self):
        self.historico_estados = [] # Pilha de histórico
        self.estado_atual = {"categoria": "Todas", "ano": 2024, "regiao": "Todas"}
        
    def aplicar_filtro(self, chave, valor):
        # Salva uma cópia do estado anterior no topo da pilha
        self.historico_estados.append(self.estado_atual.copy())
        self.estado_atual[chave] = valor
        print(f"Filtro aplicado: {chave} = {valor}")
        print("Estado atual:", self.estado_atual)
        
    def desfazer_filtro(self):
        if not self.historico_estados:
            print("Nenhum filtro para desfazer!")
            return
        # Desempilha o estado anterior
        self.estado_atual = self.historico_estados.pop()
        print("Filtro desfeito!")
        print("Estado atual revertido:", self.estado_atual)

# Teste do Projeto 1
dash = DashboardFiltros()
dash.aplicar_filtro("categoria", "Eletrônicos")
dash.aplicar_filtro("regiao", "Sudeste")
dash.desfazer_filtro()
dash.desfazer_filtro()
```

### Projeto 2: Calculadora de Expressões Aritméticas (Avaliador Infixo)
**Contexto:** Avaliar expressões infixas com pilhas de operadores e operandos.

```python
def aplicar_operador(operadores, valores):
    operador = operadores.pop()
    dir_val = valores.pop()
    esq_val = valores.pop()
    if operador == '+': valores.append(esq_val + dir_val)
    elif operador == '-': valores.append(esq_val - dir_val)
    elif operador == '*': valores.append(esq_val * dir_val)
    elif operador == '/': valores.append(esq_val / dir_val)

def precedencia(op):
    if op in ('+', '-'): return 1
    if op in ('*', '/'): return 2
    return 0

def avaliar_expressao(expressao):
    valores = []
    operadores = []
    i = 0
    while i < len(expressao):
        if expressao[i] == ' ':
            i += 1
            continue
        if expressao[i] == '(':
            operadores.append(expressao[i])
        elif expressao[i].isdigit():
            val = 0
            while i < len(expressao) and expressao[i].isdigit():
                val = (val * 10) + int(expressao[i])
                i += 1
            valores.append(val)
            i -= 1
        elif expressao[i] == ')':
            while operadores and operadores[-1] != '(':
                aplicar_operador(operadores, valores)
            operadores.pop()
        else:
            while operadores and precedencia(operadores[-1]) >= precedencia(expressao[i]):
                aplicar_operador(operadores, valores)
            operadores.append(expressao[i])
        i += 1
        
    while operadores:
        aplicar_operador(operadores, valores)
        
    return valores[-1]

# Teste do Projeto 2
expr = "3 + 4 * ( 2 - 1 )"
print(f"Resultado de '{expr}':", avaliar_expressao(expr))
```

### Exercício 2

A lista de entrada tem $n$ elementos. Qual a ordem de complexidade em função de $n$?

````python
# Exercício 2
def encontrar_maximo(lista):
    valor_maximo = lista[0]
    for numero in lista:
        if numero > valor_maximo:
            valor_maximo = numero
    return valor_maximo

# Exemplo de teste
numeros = [5, 12, 3, 19, 7]
print("Maior valor:", encontrar_maximo(numeros))

# Resposta / Análise de Complexidade:
# A complexidade de tempo é O(n) - Tempo Linear.
# O algoritmo percorre a lista de n elementos exatamente uma vez realizando comparações simples.
````

### Exercício 3
**Enunciado:** Qual a complexidade da busca binária? Considere que temos uma lista ordenada de tamanho $n$ e a função abaixo procura um elemento alvo nessa lista:
```python
def busca_binaria(lista, alvo):
    esquerda = 0
    direita = len(lista) - 1
    while esquerda <= direita:
        meio = (esquerda + direita) // 2
        if lista[meio] == alvo:
            return meio
        elif lista[meio] < alvo:
            esquerda = meio + 1
        else:
            direita = meio - 1
    return -1
```
Qual é a ordem de complexidade em função de $n$?

````python
# Exercício 3
def busca_binaria(lista, alvo):
    esquerda = 0
    direita = len(lista) - 1
    while esquerda <= direita:
        meio = (esquerda + direita) // 2
        if lista[meio] == alvo:
            return meio
        elif lista[meio] < alvo:
            esquerda = meio + 1
        else:
            direita = meio - 1
    return -1

# Exemplo de teste
lista_ordenada = [10, 20, 30, 40, 50, 60, 70]
print("Índice do alvo 40:", busca_binaria(lista_ordenada, 40))

# Resposta / Análise de Complexidade:
# A complexidade de tempo é O(log n) - Tempo Logarítmico.
# A cada iteração do laço while, o espaço de busca é reduzido pela metade.
````

### Exercício 4
**Enunciado:** Considere duas implementações que verificam se uma lista `idades` contém pelo menos duas ocorrências da menor idade presente:

```python
def existe_2_menores_v1(idades):
    tamanho = len(idades)
    menor = 200
    for i in range(tamanho):
        if idades[i] < menor:
            menor = idades[i]
    cont = 0
    for i in range(tamanho):
        if idades[i] == menor:
            cont += 1
    return cont > 1

def existe_2_menores_v2(idades):
    idades.sort()
    return idades[0] == idades[1]
```
Qual das duas funções tem menor complexidade de tempo? Justifique indicando as ordens de complexidade de cada uma.

````python
# Exercício 4
def existe_2_menores_v1(idades):
    tamanho = len(idades)
    menor = 200
    for i in range(tamanho):
        if idades[i] < menor:
            menor = idades[i]
    cont = 0
    for i in range(tamanho):
        if idades[i] == menor:
            cont += 1
    return cont > 1

def existe_2_menores_v2(idades):
    idades_copia = idades.copy() # Copia para não alterar a lista original nos testes
    idades_copia.sort()
    return idades_copia[0] == idades_copia[1]

# Exemplo de teste
idades = [25, 18, 30, 18, 40]
print("v1:", existe_2_menores_v1(idades))
print("v2:", existe_2_menores_v2(idades))

# Resposta / Análise de Complexidade:
# 1. existe_2_menores_v1:
#    - Faz dois laços simples independentes de tamanho n.
#    - Complexidade: O(n) + O(n) = O(n) (Linear).
#
# 2. existe_2_menores_v2:
#    - Ordena a lista usando .sort() (Timsort em Python), que tem complexidade O(n log n).
#    - Em seguida acessa dois elementos em O(1).
#    - Complexidade: O(n log n) (Linearítmica).
#
# Conclusão: A função 'existe_2_menores_v1' tem MENOR complexidade de tempo (O(n) < O(n log n)).
````

### Exercício 5
**Enunciado:** A função abaixo calcula o $n$-ésimo número de Fibonacci de forma recursiva:
```python
def fib(n):
    if n < 2:
        return n
    else:
        return fib(n-1) + fib(n-2)
```
Qual é a complexidade de tempo dessa implementação em termos de $n$?

````python
# Exercício 5
def fib(n):
    if n < 2:
        return n
    else:
        return fib(n - 1) + fib(n - 2)

print("Fibonacci de 6:", fib(6))

# Resposta / Análise de Complexidade:
# A complexidade de tempo é O(2^n) - Tempo Exponencial (mais precisamente O(1.618^n)).
# A cada chamada, a função se divide em duas novas chamadas recursivas, gerando uma árvore
# de chamadas de profundidade n e aproximadamente 2^n nós no total.
````

### Projeto 1: Análise de Desempenho de Busca em Dados de Clientes (Ex-Alunos)
**Contexto:** Buscar ex-alunos da Fatec Votorantim por CPF.
**Implementações:**
1. `buscar_aluno_linear`: percorre a lista de ex-alunos $O(n)$.
2. `buscar_aluno_eficiente`: mapeia ex-alunos em um dicionário usando CPF como chave $O(1)$ de tempo de busca.

````python
# Projeto 1: Análise de Desempenho de Busca em Dados de Clientes

# Dados dos ex-alunos (Lista de Dicionários)
ex_alunos = [
    {"cpf": "111.222.333-04", "nome": "Ana Souza", "curso": "Análise e Desenvolvimento de Sistemas"},
    {"cpf": "222.333.444-05", "nome": "Bruno Lima", "curso": "Gestão Empresarial"},
    {"cpf": "333.444.555-06", "nome": "Carla Dias", "curso": "Controle de Obras"},
    {"cpf": "444.555.666-07", "nome": "Daniel Alves", "curso": "Análise e Desenvolvimento de Sistemas"}
]

# 1. Busca Linear - O(n)
def buscar_aluno_linear(lista_alunos, cpf_buscado):
    for aluno in lista_alunos:
        if aluno["cpf"] == cpf_buscado:
            return aluno
    return None

# 2. Busca Eficiente - Pré-processamento em Dicionário O(n) + Busca O(1)
def criar_indice_cpf(lista_alunos):
    # Dicionário indexado por CPF: O(n) para construir
    return {aluno["cpf"]: aluno for aluno in lista_alunos}

def buscar_aluno_eficiente(dicionario_alunos, cpf_buscado):
    # Acesso direto pela chave: O(1)
    return dicionario_alunos.get(cpf_buscado, None)

# --- Teste e Demonstração ---
cpf_alvo = "333.444.555-06"

# Teste Busca Linear
resultado_linear = buscar_aluno_linear(ex_alunos, cpf_alvo)
print("Busca Linear:", resultado_linear)

# Teste Busca Eficiente
indice_alunos = criar_indice_cpf(ex_alunos) # Pré-processamento
resultado_eficiente = buscar_aluno_eficiente(indice_alunos, cpf_alvo)
print("Busca Eficiente (Dicionário):", resultado_eficiente)

# Análise de Complexidade:
# - buscar_aluno_linear: Complexidade O(n) em tempo. Cada busca percorre até n elementos.
# - buscar_aluno_eficiente: Construção da tabela hash/dicionário toma O(n) de tempo e memória. 
#   Porém, cada busca subsequente leva O(1) de tempo (constante). Ideal para sistemas de consulta frequente.
````

### Projeto 2: Recomendação de Disciplinas Baseada em Interesses
**Contexto:** Sistema de recomendação de disciplinas optativas na Fatec Votorantim baseado no histórico de alunos com disciplinas em comum.

````python
# Projeto 2: Recomendação de Disciplinas Baseada em Interesses

def recomendar_disciplinas(historicos_alunos, historico_aluno_alvo):
    """
    historicos_alunos: lista de listas, onde cada sublista contém os códigos das disciplinas cursadas por um aluno.
    historico_aluno_alvo: lista de códigos das disciplinas cursadas pelo aluno de interesse.
    """
    set_alvo = set(historico_aluno_alvo)
    recomendacoes = set()
    
    for historico_outro in historicos_alunos:
        set_outro = set(historico_outro)
        
        # Verifica se há pelo menos uma disciplina em comum
        if set_alvo.intersection(set_outro):
            # Disciplinas do outro aluno que o aluno alvo ainda NÃO cursou
            novas_disciplinas = set_outro - set_alvo
            recomendacoes.update(novas_disciplinas)
            
    return list(recomendacoes)

# --- Teste do Projeto 2 ---
historicos_outros = [
    ["ALG101", "EDD201", "BDD301"],      # Aluno A
    ["EST102", "ENG202"],               # Aluno B (Sem disciplinas em comum)
    ["ALG101", "WEB302", "IA401"],       # Aluno C (Tem ALG101 em comum)
    ["BDD301", "SEG402"]                # Aluno D (Tem BDD301 em comum)
]

historico_meu = ["ALG101", "BDD301"]

recomendacoes = recomendar_disciplinas(historicos_outros, historico_meu)
print("Disciplinas do Aluno:", historico_meu)
print("Disciplinas Recomendadas:", recomendacoes)
````

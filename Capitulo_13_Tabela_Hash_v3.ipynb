```python
def hash_string(key, m):
    s = 0
    for c in key:
        s += ord(c)
    return s % m

table_size = 7
print(hash_string("gumbo", table_size))
```

```python
class HashMap:
    def __init__(self, m=10):
        self.table = [None] * m

    def _hash(self, key):
        return hash(key) % len(self.table)

    def insert(self, key, value):
        idx = self._hash(key)
        start = idx
        while self.table[idx] is not None:
            k, _ = self.table[idx]
            if k == key:
                self.table[idx] = (key, value)
                return
            idx = (idx + 1) % len(self.table)
            if idx == start:
                raise OverflowError("Tabela hash cheia")
        self.table[idx] = (key, value)

    def search(self, key):
        idx = self._hash(key)
        start = idx
        while self.table[idx] is not None:
            k, v = self.table[idx]
            if k == key:
                return v
            idx = (idx + 1) % len(self.table)
            if idx == start:
                break
        return None

m = HashMap(5)
m.insert("João", 20)
m.insert("Ana", 30)
m.insert("Pedro", 25)
print(m.search("Ana"))
print(m.search("Maria"))
```

```python
produtos = {
    "SKU123": {"nome": "Cafeteira", "preco": 299.90},
    "SKU456": {"nome": "Mixer", "preco": 149.50},
    "SKU789": {"nome": "Batedeira", "preco": 199.00}
}

def buscar_produto(catalogo, codigo):
    return catalogo.get(codigo)

codigo = "SKU456"
info = buscar_produto(produtos, codigo)

if info is not None:
    print(f"Produto: {info['nome']} - R$ {info['preco']:.2f}")
else:
    print("Produto não encontrado")

codigo = "SKU999"
info = buscar_produto(produtos, codigo)
print(info)
```

```python
inventario = {
    "Espada de Fogo": 1,
    "Poção de Vida": 5,
    "Chave de Ouro": 2
}

def adicionar_item(invent, item, quantidade=1):
    invent[item] = invent.get(item, 0) + quantidade

def consultar_item(invent, item):
    return invent.get(item, 0)

def remover_item(invent, item, quantidade=1):
    if item not in invent:
        return False
    invent[item] -= quantidade
    if invent[item] <= 0:
        del invent[item]
    return True

adicionar_item(inventario, "Poção de Vida", 2)
print(consultar_item(inventario, "Poção de Vida"))
remover_item(inventario, "Chave de Ouro", 1)
print(inventario)
```
